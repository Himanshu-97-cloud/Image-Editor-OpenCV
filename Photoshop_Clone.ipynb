```python
# Imporring Libraries
import numpy as np
import cv2
```

```python
# Input from user of image location
user_input = input("Enter image location :")

# Reading and loading the image 
image = cv2.imread(user_input)
image
```

```python
image.shape
```

```python
def img_resize(w, h):
    resized = cv2.resize(image, (w,h))
    cv2.imshow("Resized Image", resized)
    cv2.waitKey(0)
    cv2.destroyAllWindows()

def img_crop(startY, endY, startX, endX):
    cropped =  image[startY:endY, startX:endX]
    cv2.imshow("Cropped Image", cropped)
    cv2.waitKey(0)
    cv2.destroyAllWindows()

def img_rotate(angle):
    height, width = image.shape[:2]
    center = (width//2, height//2)
    scale = 1

    M = cv2.getRotationMatrix2D(center, angle, scale)
    rotated = cv2.warpAffine(image, M, (width, height))
    cv2.imshow("Rotated Image", rotated)
    cv2.waitKey(0)
    cv2.destroyAllWindows()

def img_flip(num):
    if num == 1:
        flipped_img = cv2.flip(image, 1)
    elif num == 0:
        flipped_img = cv2.flip(image, 0)
    elif num == -1:
        flipped_img = cv2.flip(image, -1)
    else:
        print("Invalid input")

    cv2.imshow("Flipped Image", flipped_img)
    cv2.waitKey(0)
    cv2.destroyAllWindows()

def img_bright(num):
    brightness = cv2.convertScaleAbs(image, alpha=1.0, beta=num)
    cv2.imshow("Brightness of Image", brightness)
    cv2.waitKey(0)
    cv2.destroyAllWindows()

def img_contrast(num):
    brightness = cv2.convertScaleAbs(image, alpha=num, beta=0)
    cv2.imshow("Contrast of Image", brightness)
    cv2.waitKey(0)
    cv2.destroyAllWindows()

def img_grayscale():
    gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    cv2.imshow("GrayScale Image", gray)
    cv2.waitKey(0)
    cv2.destroyAllWindows()

def img_hsv():
    hsv = cv2.cvtColor(image, cv2.COLOR_BGR2HSV)
    cv2.imshow("HSV Image", hsv)
    cv2.waitKey(0)
    cv2.destroyAllWindows()

def img_save(str):
    cv2.imwrite(str, image)
    
```

```python
while True:
    print("\n1:Resize \n2:Crop \n3:Rotate \n4:Flip \n5:Brightness \n6:Contrast \n7:GrayScale \n8:HSV \n9:Save Image \n10:Exit \n")
    user_action = int(input("Enter action you wanna perform on image :"))

    match user_action:
        case 1:
            print(f"Image dimensions are : {image.shape[:2]}")
            user_width, user_height = tuple(map(int, input("Enter (width , height) of image :").split(',')))
            img_resize(user_width, user_height)
            
        case 2:
            startY, endY = tuple(map(int, input("Enter coordinates of 'Y' to crop (startY , endY) : ").split(',')))
            startX, endX = tuple(map(int, input("Enter coordinates of 'X' to crop (startX , endX) : ").split(',')))
            img_crop(startY, endY, startX, endX)

        case 3:
            user_angle = int(input("Enter angle of image to rotate from  center :"))
            img_rotate(user_angle)

        case 4:
            user_flip = int(input("Enter 1:Horizontal flip \n0:Vertical Flip \n-1:Both side Flip"))
            img_flip(user_flip)

        case 5:
            user_brighness = int(input("Enter +ve / -ve values for brightness :"))
            img_bright(user_brighness)

        case 6:
            user_contrast = float(input("Enter value 'X' (x > 1.0 > x) :"))
            img_contrast(user_contrast)

        case 7:
            img_grayscale()

        case 8:
            img_hsv()

        case 9:
            user_save = input("Enter valid format name of image to save :")
            img_save(user_save)
            
        case 10:
            print("Exiting User Actions...")
            break
```
